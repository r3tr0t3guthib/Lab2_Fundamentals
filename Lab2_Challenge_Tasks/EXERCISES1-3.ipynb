```python
import random
import numpy as np
import pandas as pd

# USER CONFIGURATION

SURNAME = "ARAGON" 
SEED_NUM = 4 

unique_seed = SEED_NUM + sum(ord(n) for n in SURNAME)
random.seed (unique_seed)
np.random.seed(unique_seed)

print(f"Unique seed generated for  {SURNAME}: {unique_seed}")
```

```python
def generate_sensor_data(num_readings=50):
    # Generates a fixed list of sensor readings based on the unique seed
    
    readings = []
    for _ in range (num_readings):
        #10% chance to generate an invalid id/corrupted reading reading for validation testing
        if random.random() < 0.10:
            invalid_type = random.choice (["string", "negative", "none"])
            if invalid_type == "string":
                readings.append ("ERROR_OOR")
            elif invalid_type == "none":
                readings.append (None)
            else:
                readings.append(random.uniform(-50.0, -1.0))

        else:
            readings.append(random.uniform(0.0, 100.0))

    return readings

raw_data = generate_sensor_data(50)
print(f"Generated {len(raw_data)} raw readings.")
```

```python
def validate_readings(raw_readings):
    """Filters out strings, None values, and negative anomalies."""
    valid_readings = []
    invalid_count = 0
    
    for val in raw_readings:
        if val is None:
            invalid_count += 1
            continue
        try:
            numeric_val = float(val)
            if numeric_val >= 0:  # Sensor baseline cannot be negative
                valid_readings.append(numeric_val)
            else:
                invalid_count += 1
        except (ValueError, TypeError):
            invalid_count += 1
            
    print(f"Validation Complete: {len(valid_readings)} Valid, {invalid_count} Invalid removed.")
    return valid_readings

# Test the validation
cleaned_data = validate_readings(raw_data)
```

```python
def classify_readings(valid_readings):
    """Classifies readings into specific operating ranges."""
    classifications = {
        "Low Operating Range (0 - 30)": 0,
        "Normal Operating Range (31 - 75)": 0,
        "Critical High Range (76+)": 0
    }
    
    for val in valid_readings:
        if val <= 30.0:
            classifications["Low Operating Range (0 - 30)"] += 1
        elif val <= 75.0:
            classifications["Normal Operating Range (31 - 75)"] += 1
        else:
            classifications["Critical High Range (76+)"] += 1
            
    return classifications

# Test the classification
thorough_counts = classify_readings(cleaned_data)
print(thorough_counts)
```

```python
def display_summary(raw, cleaned, classifications):
    """Displays a formatted analytics report of the sensor run."""
    print("=" * 45)
    print("           SENSOR DATA ANALYSIS REPORT        ")
    print("=" * 45)
    print(f"Total Raw Readings Processed: {len(raw)}")
    print(f"Successfully Validated:       {len(cleaned)}")
    print(f"Anomalies Blocked:            {len(raw) - len(cleaned)}")
    print("-" * 45)
    
    if cleaned:
        print(f"Highest Reading Recorded:     {max(cleaned):.2f}")
        print(f"Lowest Reading Recorded:      {min(cleaned):.2f}")
        print(f"Average System State:         {np.mean(cleaned):.2f}")
    print("-" * 45)
    print("OPERATING RANGE DISTRIBUTION:")
    for range_name, count in classifications.items():
        print(f" * {range_name}: {count} readings")
    print("=" * 45)

# Execute the final reporting function
display_summary(raw_data, cleaned_data, thorough_counts)
```
